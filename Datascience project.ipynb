## Problem Understanding

## Business Problem

Financial institutions receive a large number of loan applications every day. Evaluating each application manually can be time-consuming and may lead to inconsistent decisions. Approving loans for applicants who are unable to repay increases financial risk, while rejecting eligible applicants may result in lost business opportunities.

This project aims to analyze applicant information and build a Machine Learning model that predicts whether a loan application should be approved or rejected. The prediction will help financial institutions make faster, more accurate, and data-driven loan approval decisions.

## 2. Identify the target variable:

## Target Variable

The target variable in this dataset is **loan_status**.

It represents the final decision on a loan application.

Possible values:
- Approved
- Rejected

The Machine Learning model will learn from the input features to predict the value of the target variable.

##	Decide whether it is: 

## Problem Type

This project is a **Supervised Machine Learning - Classification** problem.

Reason:
- The target variable (loan_status) contains categorical values.
- The model predicts one of two classes:
    - Approved
    - Rejected

Therefore, Classification is the appropriate Machine Learning technique.

Regression is not suitable because the target variable is not continuous.

Clustering is not suitable because the dataset already contains a target variable.

### Create the Streamlit application

```python
import streamlit as st
import pandas as pd
import joblib

# Load model
model = joblib.load("loan_model.pkl")

# Load scaler
scaler = joblib.load("scaler.pkl")

st.title("Loan Approval Prediction")

st.write("Enter applicant details")

# Input fields
loan_id = st.number_input("Loan ID")
no_of_dependents = st.number_input("Number of Dependents", min_value=0)

education = st.selectbox(
    "Education",
    ["Graduate", "Not Graduate"]
)

self_employed = st.selectbox(
    "Self Employed",
    ["Yes", "No"]
)

income_annum = st.number_input("Annual Income")
loan_amount = st.number_input("Loan Amount")
loan_term_years = st.number_input("Loan Term (Years)")
cibil_score = st.number_input("CIBIL Score", min_value=300, max_value=900)

residential_assets_value = st.number_input(
    "Residential Assets Value"
)

commercial_assets_value = st.number_input(
    "Commercial Assets Value"
)

luxury_assets_value = st.number_input(
    "Luxury Assets Value"
)

bank_balance = st.number_input(
    "Bank Balance"
)

if st.button("Predict Loan Status"):

    # Create dataframe
    input_data = pd.DataFrame({
        "loan_id": [loan_id],
        "no_of_dependents": [no_of_dependents],
        "education": [education],
        "self_employed": [self_employed],
        "income_annum": [income_annum],
        "loan_amount": [loan_amount],
        "loan_term_years": [loan_term_years],
        "cibil_score": [cibil_score],
        "residential_assets_value": [residential_assets_value],
        "commercial_assets_value": [commercial_assets_value],
        "luxury_assets_value": [luxury_assets_value],
        "bank_balance": [bank_balance]
    })

    # Convert categorical values
    input_data["education"] = input_data["education"].map({
        "Graduate": 0,
        "Not Graduate": 1
    })

    input_data["self_employed"] = input_data["self_employed"].map({
        "Yes": 1,
        "No": 0
    })

    # Scale numerical columns
    scale_columns = [
        "income_annum",
        "loan_amount",
        "cibil_score",
        "residential_assets_value",
        "commercial_assets_value",
        "luxury_assets_value",
        "bank_balance"
    ]

    input_data[scale_columns] = scaler.transform(
        input_data[scale_columns]
    )

    prediction = model.predict(input_data)

    if prediction[0] == 1:
        st.success("Loan Approved")
    else:
        st.error("Loan Rejected")
```
